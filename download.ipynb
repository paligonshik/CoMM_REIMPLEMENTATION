```python
import os
import json
from glob import glob

# Define paths
dataset_dir = "1/mmimdb/dataset"
images_dir = os.path.join(dataset_dir,)
jsons_dir = os.path.join(dataset_dir)
split_file = "1/mmimdb/split.json"

# Load split.json
with open(split_file, "r") as f:
    splits = json.load(f)
def process_split(split_name):
    # Define paths
    dataset_dir = "1/mmimdb/dataset"
    images_dir = os.path.join(dataset_dir,)
    jsons_dir = os.path.join(dataset_dir)
    split_file = "1/mmimdb/split.json"

    # Load split.json
    with open(split_file, "r") as f:
        splits = json.load(f)

    # Prepare list for dataset entries and a set for unique genres
    dataset_entries = []
    all_genres = set()

    # Let's focus on the specified split for collecting genres
    for movie_id in splits[split_name]:
        json_path = os.path.join(jsons_dir, f"{movie_id}.json")
        image_path = os.path.join(images_dir, f"{movie_id}.jpeg")

        # Load JSON file
        with open(json_path, "r") as jf:
            data = json.load(jf)

        # Extract plot (join if necessary)
        plot = " ".join(data.get("plot", []))

        # Extract genres (list of strings)
        genres = data.get("genres", [])

        # Check if the movie should be excluded based on its genres
        if any(genre in ['News', 'Reality-TV', 
                         'Talk-Show',"Adult"] for genre in genres):
            continue

        # Update the unique genre set
        for genre in genres:
            all_genres.add(genre)

        # Create a dictionary for this movie
        entry = {
            "id": movie_id,
            "image_path": image_path,
            "plot": plot,
            "genres": genres
        }
        dataset_entries.append(entry)

    # Save the dataset entries to a JSON file for later use
    with open(f"{split_name}.json", "w") as out_f:
        json.dump(dataset_entries, out_f, indent=2)

    return all_genres

# Example usage:
train_genres = process_split("train")
test_genres = process_split('test')
dev_genres = process_split("dev")
```
